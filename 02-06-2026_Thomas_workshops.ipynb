```python
# 07 Ranges  - 05 Kontrollstrukturen
# Aufgabe 1 - Ausgabe der ersten n Quadratzahlen

def print_squares(n: int):
    """ 
    Takes an integer n and forms a list with the first n integers.
    Prints every number of the list with its square.
    """
    for number in list(range(n+1)):
        print(f"{number}^2 = {(number)**2}")
```

```python
print_squares(15)
```

```python
# 07 Ranges - 05 Kontrollstrukturen
# Aufgabe 2 - Ausgabe der Summe der ersten n Quadratzahlen

def sum_squares(n):
    """ 
    Takes an integer n and forms a list of the first n integers of it.
    Every element of that list is squared and then added to a final result.
    Function returns the sum of the squares of the first n integers.
    """
    result = 0
    for number in list(range(n+1)):
        square = (number)**2
        result += square
    return result

```

```python
def count_unique_words(text: str):
    """ 
    Takes the input string and writes it in lowercase. 
    Generates a list with every substring in it, 
    set() transforms this into a set, so every string is unique. 
    We then substract the punctoation marks "," and "." and count the remaining strings.
    """
    return len(set(text.lower().split()) - {",", "."})
```

```python
# 01 Dictionaries - 07 Weitere Datenstrukturen

# Workshop 1 - Secondhand Verkauf
# ---------------------------------------------
# Sie beschließen nicht mehr benötigte Kleidung bei einem Secondhand-Shop zu verkaufen. 
# Um zu berechnen, wie hoch die Einnahmen sein werden, schreiben Sie eine Funktion gesamtpreis(kleidungsstücke: list, preise: dict) -> float:


alte_kleidung = ["Wintermantel", "Cordhose", "Smoking"]
secondhand_preise = {"Wintermantel": 80, "Smoking": 40} 

def gesamtpreis(kleidungsstuecke: list, preise: dict):
    """
    Loops through every item in argument 1 and looks for a corresponding key in argument 2.
    If the item is found it adds the value to preis, is nothing is found, 0 will be added.
    Returns the total sum. 
    """
    preis = 0.0
    for stuff in kleidungsstuecke:
        preis += preise.get(stuff, 0)
    return preis
    
```

```python
# 01 Dictionaries - 07 Weitere Datenstrukturen

# Workshop 2 - Be happy :)
# ---------------------------------------------
# Schreiben Sie eine Funktion `replace_words(text: str, replacements: dict)`, die alle
# Wörter, die in `dict` als Key vorkommen durch ihren Wert in `dict` ersetzen.

advice = "Don't worry be happy"
smileys = {"worry": "\U0001f61f", "happy": "\U0001f600"}

def replace_words(text: str, replacements: dict):
    """ 
    Takes a string and splits it to a list. Loops through this list and compares it with the dictionary. 
    If a key is found the corresponding value will be put to a new empty list, else the element itself will be added to the new empty list.
    From this new list a string is generated and returned.
    """
    words = text.split()
    new_words = []
    
    for word in words:
        if word in replacements.keys():
            new_words.append(replacements[word])
        else:
            new_words.append(word)
    string = " ".join(new_words)
    return string
```

```python
replace_words(advice,smileys)
```

```python
# version 2 with the get() method to avoid if-else. Otherwise the same.
def replace_words2(text: str, replacements: dict):
    words = text.split()
    new_words = []

    for word in words:
        new_words.append(replacements.get(word, word))
    string = " ".join(new_words)
    return string
```

```python
replace_words2(advice, smileys)
```

```python
# 01 Dictionaries - 07 Weitere Datenstrukturen

# Workshop 3 - Inverser Index PART I
# ---------------------------------------------
# Schreiben Sie eine Funktion `first_index(a_list: list) -> dict`, die ein Dictionary
# `d` zurückgibt, für das gilt:

# - Die Schlüssel in `d` sind alle in `a_list` vorkommenden Werte
# - Für jeden Wert `elt`, der in `a_list` enthalten ist, ist der Wert von `d[elt]`
#  der erste Index, an dem `elt` in `a_list` vorkommt


def first_index(a_list: list):
    dic = {}
    for index, element in enumerate(a_list):
        dic.setdefault(element, index)
    return dic
```

```python
# 01 Dictionaries - 07 Weitere Datenstrukturen

# Workshop 3 - Inverser Index PART II
# ---------------------------------------------
# Schreiben Sie eine Funktion `inverse_index(a_list: list) -> dict`, die ein Dictionary
# `d` zurückgibt, für das gilt:

# - Die Schlüssel in `d` sind alle in `a_list` vorkommenden Werte
# - Für jeden Wert `elt`, der in `a_list` enthalten ist, ist der Wert von `d[elt]`
#  die Liste mit allen Indizes, an dem `elt` in `a_list` vorkommt

def inverse_index(a_list: list):
    """
    We loop through the indices and elements of a list.
    Every element will be a key in a dictionary and the potentially many indices of it will be in a list-value to that key.
    """
    dic = {}
    for index, element in enumerate(a_list):
        # don't forget setdefault returns the 2. argument
        dic.setdefault(element, []).append(index)
    return dic
```

```python
# 02 Iteration über Dictionaries - 07 Weitere Datenstrukturen

# Drucken von Lagerbeständen
# ------------------------------------------------------------
# Wir schreiben die Software für ein Hochregallager, in dem der Lagerbestand in einem Dictionary verwaltet wird.
# Die Keys im Dictionary sind die Plätze im Lager, die Werte das am jeweiligen Platz gespeicherte Produkt.
# Schreiben Sie die folgenden Funktionen:
#   print_inventory(warehouse), die alle im Lager vorhandenen Produkte ausdruckt
#   print_locations(warehouse), die alle Plätze im Lager ausdruckt
#   print_assignments(warehouse), die für jedes Produkt ausgibt, in welchem Platz es gespeichert ist.

my_warehouse = {(0, 0): "plugs", (1, 0): "cables", (0, 1): "circuit boards"}

def print_inventory(warehouse: dict):
    """Prints only the values of the dictionary."""
    for product in warehouse.values():
        print(product)

def print_locations(warehouse: dict):
    """Prints only the keys of the dictionary."""
    for location in warehouse.keys():
        print(location)

def print_assignements(warehouse: dict):
    """
    Prints both the values and the keys of the dictionary in a tabular way.
    Takes the longest value max(len(...)) as column width.
    """
    width = max(len(str(value)) for value in warehouse.values())
    for location, product in warehouse.items():
        print(f"{product:<{width}} -> {location}")
```

```python
# 08 - Klassen, Objekte und Methoden - 03 Klassen (Teil 2)

class Kfz:
    def __init__(self, hersteller = None, kennzeichen = None) -> None:
        self.hersteller = hersteller
        self.kennzeichen = kennzeichen
        

    def melde_um(self, neues_kennzeichen):
        self.kennzeichen = neues_kennzeichen
    

    def __repr__(self) -> str:
        return f"KFZ('{self.hersteller}', '{self.kennzeichen}')"
    

    def __eq__(self, other):
        if not isinstance(other, Kfz):
                return NotImplemented
        return other.hersteller == self.hersteller and other.kennzeichen == self.kennzeichen

    
```

```python
# 11 - Vererbung & Polymorphie - 02 Vererbung (Workshops) 

# Mitarbeiter

class Mitarbeiter:
    def __init__(self, name = None, personalnummer = 0, grundgehalt = 0):
        self.name = name
        self.personalnummer = personalnummer
        self.grundgehalt = grundgehalt

    def __repr__(self):
        return f"Mitarbeiter({self.name!r}, {self.personalnummer}, {self.grundgehalt})"

    def gehalt(self):
        return 13/12 * self.grundgehalt

```

```python
class Arbeiter(Mitarbeiter):
    def __init__(self, name, personalnummer, grundgehalt, ueberstunden, stundenlohn):
        super().__init__(name, personalnummer, grundgehalt)
        self.ueberstunden = ueberstunden
        self.stundenlohn = stundenlohn
    
    def __repr__(self):
        return f"Arbeiter({self.name!r}, {self.personalnummer}, {self.grundgehalt}, {self.ueberstunden}, {self.stundenlohn})"

    def gehalt(self):
        return super().gehalt() + self.ueberstunden * self.stundenlohn



class Manager(Mitarbeiter):
    def __init__(self, name, personalnummer, grundgehalt, bonus):
        super().__init__(name, personalnummer, grundgehalt)
        self.bonus = bonus

    def __repr__(self):
        return f"Manager({self.name!r}, {self.personalnummer}, {self.grundgehalt}, {self.bonus})"

    def gehalt(self):
        return super().gehalt() + self.bonus
```

```python
# 11 - Vererbung & Polymorphie - 03 Abstrakte Klassen (Teil 2)


from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self,base,height):
        ...

    @abstractmethod
    def circumference(self,base,height):
        ...
        
 
```

```python
class Square(Shape):
    def area(self,base,height=0):
        return base**2
    
    def circumference(self, base, height=0):
        return 4 * base
    

class Rectangle(Shape):
    def area(self,base,height):
        return base * height
    
    def circumference(self, base, height):
        return 2 * (base + height)
    

class Triangle(Shape):
    def area(self,base ,height):
        return 1/2 * (base * height)
    
    def circumference(self, base, height):
        return (base**2 + height**2)**0.5
    

class Drawing(Shape):
    def shapes(self):
        all_shapes = []
        total_area = 0
        total_circumference = 0

        while True:
            user_input = input(f"Which shape do you want to add to our list? (Triangle, Rectangle, Square)").lower().strip()

            if user_input == "triangle":
                all_shapes.append("Triangle")
                base = float(input("What is the base of this triangle? "))
                height = float(input("What is the height of this triangle? "))
                triangle = Triangle()
                total_area += triangle.area(base, height)
                total_circumference += triangle.circumference(base, height)

            elif user_input == "rectangle":
                all_shapes.append("Rectangle")
                base = float(input("What is the base of this Rectangle? "))
                height = float(input("What is the height of this Rectangle? "))
                rectangle = Rectangle()
                total_area += rectangle.area(base, height)
                total_circumference += rectangle.circumference(base, height)

            elif user_input == "square":
                all_shapes.append("Square")
                base = float(input("What is the base of this square? "))
                square = Square()
                total_area += square.area(base)
                total_circumference += square.circumference(base)
            
            # break condition from user
            go_on = input("Add more shapes? (j/n)")
            if go_on.lower().strip() == "n":
                break

        print((f"The total area for your shapes is: {total_area}.\n"
                f"The total circumference is: {total_circumference}.\n"
                f"This is your list of shapes: {all_shapes} "))
        
        return None
    
```

```python
# 09 Objektmodell und Magic Methods - 02 Dataclasses

# Einkaufsliste

from dataclasses import dataclass, field

@dataclass
class ShoppingListItem:
    product: str 
    price: float 
    amount: int = field(default=0)

    def total_price(self):
        return self.amount * self.price



@dataclass
class ShoppingList:
    items: list = field(default_factory=list)


    def add_item(self, new_item: ShoppingListItem):
        exists = False
        for index, old_item in enumerate(self.items):
            if old_item.product == new_item.product and old_item.price == new_item.price:
                self.items[index].amount += new_item.amount
                exists = True
        if not exists: self.items.append(new_item)
    

    def total_price(self):
        price = 0
        for item in self.items:
            price += item.amount * item.price
        return price
    

    def show_shoppinglist(self):
        print("Einkaufsliste")
        for i in range(len(self.items)):
            print(f"    {self.items[i].amount} x {self.items[i].product} à {self.items[i].price} = {self.items[i].total_price()}")
        print(f"Gesamt: {self.total_price()}")


    def __len__(self):
        return len(self.items)
    

    def __getitem__(self, index):
        if len(self.items) > index:
            return self.items[index]
        else:
            return None


        


```
