```python
# Particles on a 1D grid

# List of particles with position x and velocity v (per timestep)
particles = {"A": {"x": 0, "v": 3},
             "B": {"x": 3, "v": 4},
             "C": {"x": 10, "v": -1},
             "D": {"x": 5, "v": 2}}

```

```python
def simulate_step(particles):
    """ 
    Particles move one timestep forward, so that velocities are added to positions.
    """
    for attributes in particles.values():
        attributes['x'] += attributes['v']
    return particles
```

```python
evolved_particles = simulate_step(particles)
print(evolved_particles)
```

```python
def find_collisions(particles):
    """ 
    Returns a dictionary with positions as keys whenever at least two particles collide.
    """
    collision_pairs = {}
    positions = {}
    
    for particle, attributes in particles.items():
        positions.setdefault(attributes['x'], []).append(particle)

    for position, particle_list in positions.items():
        if len(particle_list) > 1:
            collision_pairs[position] = tuple(particle_list)

    return collision_pairs


```

```python
collision_pairs = find_collisions(particles)
print(collision_pairs)
```

```python
def elastic_collision(collision_pairs: dict, particles: dict):
    """ 
    The velocities for every colliding pair are flipped.
    """
    for pos, particle_pair in collision_pairs.items():
        p1, p2 = particle_pair # -> keys von particles
        particles[p1]['v'], particles[p2]['v'] = particles[p2]['v'], particles[p1]['v']
    return particles

```

```python
elastic_collision(collision_pairs,evolved_particles)
```
