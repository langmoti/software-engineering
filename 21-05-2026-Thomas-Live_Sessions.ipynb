## 2 Numbers Calculator Functions

```python
# Addition
def add_two_numbers(a,b):
    summe = a + b
    return summe
```

```python
# Subtraction
def subtract_two_numbers(a,b):
    difference = a - b
    return difference
```

```python
# Multiplication
def multiply_two_numbers(a,b):
    product = a * b
    return product
```

```python
# Division
def divide_two_numbers(a,b):
    if b == 0:
        print("Nicht durch Null teilen :P ")
        return
    else:
        quotient = a / b
    return quotient
```

```python
# Modulo
def mod_two_numbers(a,b):
    modulo = a % b
    return modulo
```

```python
# multiplication by calling the add_two_numbers function
def multiply(a,b):
    value = a * add_two_numbers(b,0)
    return value
```

```python
# exponentiation a^b
def exponentiation(a,b):
    # erst die edge cases b=0 und b=1
    if b == 0:
        value = 1
    elif b == 1:
        value = a
    # "echte Exponentiation" als rekursive Funktion die die Multiplikationsfunktion aufruft
    else:
        value = multiply_two_numbers(a,exponentiation(a,b-1))
    return value
    
```

```python
# Kosinussatz für allgemeine Dreiecke
def sidelength(length1, length2):
    # ask user for angle
    winkel = input(f"Welcher Winkel wird zwischen den beiden Seiten eingeschlossen: ")

    # convert angle to radians
    phi = float(winkel) * m.pi/180

    # actual cosine function and rounding to 2 decimal places
    length3 = round((length1 ** 2 + length2 ** 2 - 2*length1*length2*m.cos(phi) ) ** 0.5,2)
    return length3
```

```python
def steuerhilfe_mitgliedschaft():
    """
    Beitragsrechner für die Mitgliedschaft im Steuerhilfeverein.
    Beiträge berechnen sich nach dem Gehalt.
    ## Notiz: "Schlechte" Version weil Symmetrieprinzip verletzt wird oben > dann < und boundaries zu eng, man könnte es nur mit ifs machen
    """
    gehalt = int(input("Wie hoch ist Ihr Gehalt? "))

    # Überprüfen des Mitgliedsbeitrags
    if gehalt > 60000:
        print("Ihre Mitgliedschaft kostet 300€")
    elif 50000 <= gehalt < 60000:
        print("Ihre Mitgliedschaft kostet 250€")
    elif 40000 <= gehalt < 50000:
        print("Ihre Mitgliedschaft kostet 200")
    elif 30000 <= gehalt < 40000:
        print("Ihre Mitgliedschaft kostet 150€")
    elif 20000 <= gehalt < 30000:
        print("Ihre Mitgliedschaft kostet 100€")
    elif 15000 <= gehalt < 20000:
        print("Ihre Mitgliedschaft kostet 50€")
    else:
        print("Für Sie ist die Mitgliedschaft umsonst.")
    return gehalt

```

### List Methoden

#### Elemente hinzufügen
1. append(abc) -> fügt am Ende der Liste ein Element hinzu
2. extend(<>,<>,...) -> fügt am Ende der Liste eine Anzahl neuer Elemente hinzu
3. insert(index, abc) -> fügt an Indexposition index das Element hinzu und verschiebt alle anderen dahinter


#### Elemente entfernen
1. del(List[i]) -> löscht von der Liste das Element an Index i
2. List.pop(Element) -> löscht das Element aus der Liste und liefert es zurück
3. List.remove(Element) -> löscht das erste(!) Element mit dem Namen aus der Liste


#### Sortieren
1. List.sort() -> sortiert diese Liste aufsteigend
2. new = sorted(List) -> sortiert die Liste und gibt sie als Ergebnis zurück --> Speichern!


#### Index
- List.index(Element) -> gibt den Index des Element 
- Letztes Element:  List[len(List - 1)] <=> List[-1]

#### Count
List.count(Element) -> Gibt die Häufigkeit von Element in der Liste

```python
# returns a list of all even integers from 1 to n
def even_nums(n):
    my_nums = []
    
    while n > 0:
        if n % 2 == 0:
            my_nums.append(n)
        else:
            pass
        n -= 1
    return my_nums
```

```python
def remove_all_elements(name: str, a: list):
    # create a copy of the original list, so the list a exists further!
    copy_of_list = a.copy()
    while name in copy_of_list:
            copy_of_list.remove(name)
    return copy_of_list
```

```python
matrix = [
    [1,2,3], 
    [10,20,30], 
    [42,2,2],
    [5,7,9]
    ]
```

```python
def search_matrix_index(matrix: list, element):
    """ 
    Takes a matrix and an element to look for.
    If the the element is in the matrix, the function counts the occurences and print the index positions.
    Else it prints a message that the element is not in the matrix.
    """
    element_counter = 0
    result_list = []
    for row in range(len(matrix)):
        for column in range(len(matrix[row])):
            if element == matrix[row][column]:
                position = (row, column)
                element_counter += 1
                result_list.append(position)
    # if the element is not in the matrix, print statement
    if element_counter == 0:
        print(f"Dein gesuchtes Element {element} ist nicht in der Matrix.")
    # print the occurences and the positions
    else: 
        result_list.insert(0,element_counter)
        return result_list
```

```python
search_matrix_index(matrix, 42)
```
